```python
my_set= set(range(1,11))
print(my_set)
```

```python
my_set.add(11)
my_set.remove(1)
print(my_set)
```

```python
s1=set(range(1,6))
s2={x for x in range(1,11) if x%2==0}
print(s2)
print(s1)

union_set= s1.union(s2)
print(union_set)

difference_set= s1.difference(s2)
print(difference_set)

int_set= s1.intersection(s2)
print(int_set)

symdiff_set= s1.symmetric_difference(s2)
print(symdiff_set)

print(s2.issubset(s1))
print(s1.issuperset(s2))

print(s1.symmetric_difference_update(s2))
```
